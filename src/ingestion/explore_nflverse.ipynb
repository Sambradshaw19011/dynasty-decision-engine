```python

import nflreadpy as nfl
```

```python
def inspect_dataset(name, data):
    print("\n" + "=" * 70)
    print(f"DATASET: {name}")
    print("=" * 70)

    print(f"Rows: {data.height:,}")
    print(f"Columns: {len(data.columns)}")

    print("\nColumn names:")
    print(data.columns)

    print("\nData types:")
    print(data.schema)

    print("\nFirst 5 records:")
    print(data.head(5))


# 1. Player information
players = nfl.load_players()
inspect_dataset("PLAYERS", players)

# 2. Historical player statistics
player_stats = nfl.load_player_stats([2022, 2023, 2024, 2025])
inspect_dataset("PLAYER STATS", player_stats)

# 3. Historical rosters
rosters = nfl.load_rosters([2024, 2025])
inspect_dataset("ROSTERS", rosters)

# 4. Historical NFL draft selections
draft_picks = nfl.load_draft_picks()
inspect_dataset("DRAFT PICKS", draft_picks)

# 5. Combine results
combine = nfl.load_combine()
inspect_dataset("COMBINE", combine)

# 6. Game schedules
schedules = nfl.load_schedules([2024, 2025])
inspect_dataset("SCHEDULES", schedules)

# 7. Weekly depth charts
depth_charts = nfl.load_depth_charts([2025])
inspect_dataset("DEPTH CHARTS", depth_charts)

# 8. Next Gen Stats
ngs = nfl.load_nextgen_stats(
    stat_type="receiving",
    seasons=[2024, 2025]
)
inspect_dataset("NEXT GEN RECEIVING", ngs)
```

```python
weekly_stats = nfl.load_player_stats(
    [2022, 2023, 2024, 2025],
    summary_level="week"
)

inspect_dataset("WEEKLY PLAYER STATS", weekly_stats)
```
